```python
fig, axes = plt.subplots(3, 2, figsize=(8, 12))

plt.subplots_adjust(hspace=0, wspace=0)

ylims = [
    [(0.04, 1e2), (0.04, 1e2)],   # row 1
    [(1.8, 200), (1.8, 200)],   # row 2
    [(0.2, 100), (0.2, 100)]    # row 3
]

# Define the data sets in the desired order
datasets = [
    # Left column, right column
    [(O_1982_x_func, O_1982_f0, O_1982_f1, O_1982_f2, O_1982_x_graph, O_1982_y_graph, O_1982_err_graph, "1.982 MeV 18O"),
     (F_3062_x_func, F_3062_f0, F_3062_f1, F_3062_f2, F_3062_x_graph, F_3062_y_graph, F_3062_err_graph, "3.062 MeV 18F")],
    [(O_3552_x_func, O_3552_f0, None, None, O_3552_x_graph, O_3552_y_graph, O_3552_err_graph, "3.552 MeV 18O"),
     (F_4652_x_func, F_4652_f0, None, None, F_4652_x_graph, F_4652_y_graph, F_4652_err_graph, "4.652 MeV 18F")],
    [(O_3920_x_func, O_3920_f0, O_3920_f1, O_3920_f2, O_3920_x_graph, O_3920_y_graph, O_3920_err_graph, "3.920 MeV 18O"),
     (F_4964_x_func, F_4964_f0, F_4964_f1, F_4964_f2, F_4964_x_graph, F_4964_y_graph, F_4964_err_graph, "4.964 MeV 18F")]
]


    
for i in range(3):
    for j in range(2):
        ax = axes[i, j]
        ax.set_yscale("log")
        if j == 0:
            ax.set_xlim(5, 40)
        else:
            ax.set_xlim(5.1, 40)
            ax.tick_params(labelleft=False)
        
        if i == 0 or i == 1:
            ax.tick_params(labelbottom=False)

        ax.set_ylim(ylims[i][j])

        ax.tick_params(axis='both', which='both', direction='in', top=True, right=True)
    
for i, row in enumerate(datasets):
    for j, data in enumerate(row):
        ax = axes[i, j]
        x_func, f0, f1, f2, x_graph, y_graph, err_graph, label = data

        # Plot smooth function lines
        ax.plot(x_func, f0, label='f0', color='blue')
        if f1 is not None:
            ax.plot(x_func, f1, label='f1', color='red')
        if f2 is not None:
            ax.plot(x_func, f2, label='f1', color='green')

        # Plot graph points with error bars
        ax.errorbar(x_graph, y_graph, yerr=err_graph, fmt='o', color='black', label='Data')

        ax.set_yscale("log")
        ax.set_title(label, fontsize=10)
        ax.grid(False)

        # Only show legend in the first subplot
        if i == 0 and j == 0:
            ax.legend(fontsize=8)

# Set common axis labels
fig.text(0.5, 0.04, "Angle (deg)", ha='center', fontsize=12)
fig.text(0.04, 0.5, "Cross Section (mb/sr)", va='center', rotation='vertical', fontsize=12)

plt.show()
```
