```python
bin_width = x1[1] - x1[0]
legend_pos = (0.535, 0.59)
legend_pos1 = (0.535, 0.7)

def make_edges(x, width):
    return np.concatenate([x - width/2, [x[-1] + width/2]])

fig, axes = plt.subplots(2, 1, figsize=(10.5, 14), gridspec_kw={"height_ratios":[1,1]})
plt.subplots_adjust(hspace=0.0, left=0.14, right=0.97, top=0.95, bottom=0.12)

for ax in axes:
    ax.xaxis.set_minor_locator(MultipleLocator(2))
    ax.tick_params(which="both", direction="in", top=True, right=True)

ax = axes[0]

edges2 = make_edges(x2, bin_width)
edges1 = make_edges(x1, bin_width)

ax.step(edges2, np.append(y2, y2[-1]), where='post', color='#d62728', lw=1, label=r'$^{17}$O(d,p)')
ax.step(edges1, np.append(y1, y1[-1]), where='post', color='#2ca02c', lw=1, label=r'$^{17}$F(d,p)')

x_line = 5.6071
vline = ax.vlines(x_line, ymin=ax.get_ylim()[0], ymax=1900,
          color='black', linestyle='--', linewidth=1, label=r'$S_p$ of $^{18}$F = 5.61 MeV')

ax.set_ylabel("Counts / 70 keV", fontsize=24, labelpad=15)

ax.set_xticklabels([])
ax.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.yaxis.set_major_locator(MultipleLocator(200))
ax.yaxis.set_minor_locator(MultipleLocator(50))
ax.set_ylim((0.1, 1820))

ax.tick_params(which='major', length=8, top=False, right=False, labelsize=22)   # major ticks 8 points
ax.tick_params(which='minor', length=4, top=False, right=False)

ax.legend(fontsize=21, frameon=False, loc='lower left', bbox_to_anchor=legend_pos1, labelspacing=0.15, handlelength=0.9, handletextpad=0.3)

# bottom panel

ax = axes[1]

edges5 = make_edges(x5, bin_width)
edges3 = make_edges(x3, bin_width)
edges4 = make_edges(x4, bin_width)

ax.step(edges5, np.append(y5, y5[-1]), where='post', color='#ff7f0e', lw=1,
        label=r'$^{17}$F(d,p) with $^{18}$F recoils')
ax.step(edges3, np.append(y3, y3[-1]), where='post', color='#1f77b4', lw=1,
        label=r'$^{17}$F(d,p) with $^{17}$O recoils')
ax.step(edges4, np.append(y4, y4[-1]), where='post', color='black', lw=1, alpha=0.3,
        label=r'$^{17}$F(d,p) with $^{14}$N recoils $\times$ 7')

# Vertical lines
x_line_alpha = 4.415

vline = ax.vlines(x_line, ymin=ax.get_ylim()[0], ymax=1900,
          color='black', linestyle='--', linewidth=1, label=r'$S_p$ of $^{18}$F = 5.61 MeV')
vline2 = ax.vlines(x_line_alpha, ymin=ax.get_ylim()[0], ymax=1900,
          color='black', linestyle='--', alpha=0.8, linewidth=1, label=r'$S_\alpha$ of $^{18}$F = 4.415 MeV')

ax.set_ylabel("Counts / 70 keV", fontsize=26, labelpad=15)
ax.set_xlabel("E$_x$ (MeV)", fontsize=26, labelpad=12)

ax.legend(fontsize=21, frameon=False, loc='lower left', bbox_to_anchor=legend_pos, labelspacing=0.15, handlelength=0.9, handletextpad=0.3)
ax.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.xaxis.set_minor_locator(MultipleLocator(0.5))
ax.yaxis.set_major_locator(MultipleLocator(200))
ax.yaxis.set_minor_locator(MultipleLocator(50))
ax.set_ylim((0, 1820))

ax.tick_params(which='major', length=8, top=False, right=False, labelsize=22)   # major ticks 8 points
ax.tick_params(which='minor', length=4, top=False, right=False)


#fig.text(0.5, 0.98, "Excitation Spectra", ha='center', va='bottom', fontsize=14, fontweight='bold')

fig.savefig("ex_F_O_N_step.pdf", dpi=600, bbox_inches="tight")
plt.show()
```
