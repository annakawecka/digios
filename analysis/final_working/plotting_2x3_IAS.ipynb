```python
from scipy.stats import norm

energy_ratio_d = np.array([4.7, 4.942])
ratio_ratio_d = np.array([0.8479, 1.07])
unc_ratio_d = np.array([0.157, 0.287])

energy_ratio_s = np.array([3.062, 4.942, 6.136])
ratio_ratio_s = np.array([0.679, 0.983, 1.49])
unc_ratio_s = np.array([0.437, 0.5186, 0.42])
```

```python
fig, ax = plt.subplots(figsize=(7, 5))

mean_s = np.mean(ratio_ratio_s)
rms_s = np.std(ratio_ratio_s, ddof=1)

weights = 1 / unc_ratio_s**2
mean_s_w = np.sum(weights * ratio_ratio_s) / np.sum(weights)
rms_s_w = np.sqrt(1 / np.sum(weights))

x = ratio_ratio_s[-1]
sigma = unc_ratio_s[-1]

z = (x - 1) / sigma

p_value = 2 * (1 - norm.cdf(abs(z)))

print(p_value)

x_last = energy_ratio_s[-1]
y_last = ratio_ratio_s[-1]
sigma_last = unc_ratio_s[-1]

ax.errorbar(energy_ratio_d, ratio_ratio_d, yerr=unc_ratio_d, fmt='o', label=r"$\ell = 2$", capsize=5, color='#1f77b4')
ax.errorbar(energy_ratio_s, ratio_ratio_s, yerr=unc_ratio_s, fmt='s', label=r"$\ell = 0$", capsize=5, color="#ff7f0e")

vline = ax.vlines(x_line, ymin=0.1, ymax=2.2,
          color='black', linestyle='--', linewidth=1, label=r'$S_p$ of $^{18}$F = 5.61 MeV')

ax.set_xlabel('Energy (MeV)')
ax.set_ylabel('SF Ratio')
ax.set_title('Energy vs Ratio with Uncertainties')
ax.grid(False)

ax.hlines(mean_s, xmin=2.5, xmax=6.9, colors='gray', linestyles='-', label=r'Mean ($\ell$ = 0)')

ax.fill_betweenx(y=[mean_s - rms_s, mean_s + rms_s], x1=2.5, x2=6.9, color='gray', alpha=0.2, label=r'Mean $\pm$ 1$\sigma$ ($\ell$ = 0)')

ax.set_ylim((0.1, 2.2))
ax.set_xlim((2.5, 6.9))

ax.annotate(f'p = {p_value:.2f}',
    xy=(x_last, y_last),
    xytext=(x_last + 0.2, y_last + 0.2),
    arrowprops=dict(arrowstyle='->')
)

ax.legend(frameon=False)

fig.savefig("ratio_SF.pdf", dpi=600, bbox_inches="tight")
plt.show()
```
