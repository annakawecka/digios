```python
offset_O = 1.04155

fig, ax = plt.subplots(figsize=(16*2,12*2))

# Plot 18F levels
for e, l, pl in zip(E_18F, labels_18F, labels_18F_placement):
    if e is not None:
        ax.hlines(e, 0.9, 1.3, colors='black', linewidth=4)
        if l:
            ax.text(0.85, pl, l, ha='right', va='center', fontsize=42)

# Plot 18O levels
for e, l, pl in zip(E_18O, labels_18O, labels_18O_placement):
    if e is not None:
        ax.hlines(e + offset_O, 1.7, 2.1, colors='black', linewidth=4)
        if l:
            ax.text(2.15, pl + offset_O, l, ha='left', va='center', fontsize=42)

# Connect IAS levels
for eF, eO in zip(E_18F, E_18O):
    if eO is not None:
        ax.plot([1.3, 1.7], [eF, eO+ offset_O], 'k--', linewidth=2)

# Styling
ax.set_xticks([1, 2])
ax.set_xticklabels(['$^{18}$F', '$^{18}$O'], fontsize=64)
ax.tick_params(axis='x', length=0)

ax.set_xlim(0.5, 2.5)
ax.set_ylim(-0.2, 7)
ax.set_yticks([])                   # no y ticks
ax.set_ylabel("")

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)

arrow_x = 0.55  # x-position for the arrow
ax.annotate(
    "", 
    xy=(arrow_x, ax.get_ylim()[1] - 0.2), 
    xytext=(arrow_x, ax.get_ylim()[0] + 0.2),
    arrowprops=dict(
        arrowstyle="-|>,head_length=2.,head_width=1", 
        lw=5, 
        color="black"
    )
)
ax.text(
    arrow_x - 0.05, 
    (ax.get_ylim()[1] + ax.get_ylim()[0]) / 2, 
    "Energy", 
    rotation=90, 
    va="center", 
    ha="center", 
    fontsize=56
)

plt.tight_layout()
plt.savefig("energy_levels.png", dpi=600, bbox_inches="tight") 
plt.show()
```

```python
from matplotlib.patches import Rectangle

# Example data: (Z, N, symbol, mass, half-life)
# Half-life strings can be formatted however you like (s, min, h, y)
isotopes = [
    (7, 8, "N", 15, "Stable"),
    (7, 9, "N", 16, "7.1 s"),
    (7,10, "N", 17, "4.2 s"),
    (7,11, "N", 18, "619 ms"),
    (7,12, "N", 19, "336 ms"),

    (8, 8, "O", 16, "Stable"),
    (8, 9, "O", 17, "Stable"),
    (8,10, "O", 18, "Stable"),     # 18O of interest
    (8,11, "O", 19, "26.5 s"),
    (8,12, "O", 20, "13.5 s"),

    (9, 8, "F", 17, "64.4 s"),
    (9, 9, "F", 18, "109.8 min"),  # 18F of interest
    (9,10, "F", 19, "Stable"),
    (9,11, "F", 20, "11.0 s"),
    (9,12, "F", 21, "4.2 s"),

    (10, 8, "Ne", 18, "1.7 s"),
    (10, 9, "Ne", 19, "17.3 s"),
    (10, 10, "Ne", 20, "Stable"),
    (10, 11, "Ne", 21, "Stable"),
    (10, 12, "Ne", 22, "Stable")
]

# Highlight nuclei of interest
highlight = {"18O": "#FF887E", "18F": "#95C8FF"}

fig, ax = plt.subplots(figsize=(8, 8), dpi=150)

for Z, N, sym, mass, halflife in isotopes:
    x = N
    y = Z
    name = f"{mass}{sym}"
    color = "white"
    edgecolor = "black"

    if name in highlight:
        color = highlight[name]
        edgecolor = "black"

    # Draw rectangle for isotope
    rect = Rectangle((x-0.5, y-0.5), 1, 1,
                     facecolor=color, edgecolor=edgecolor, lw=1.5)
    ax.add_patch(rect)

    # Add isotope name
    ax.text(x, y+0.15, name, ha="center", va="center", fontsize=18, weight="bold")
    # Add half-life
    ax.text(x, y-0.2, halflife, ha="center", va="center", fontsize=14)

# Axis labels
ax.set_xlabel("Neutron Number (N)", fontsize=18)
ax.set_ylabel("Proton Number (Z)", fontsize=18)

ax.set_yticks(range(7, 11))

ax.tick_params(axis='x', labelsize=14)
ax.tick_params(axis='y', labelsize=14)

ax.set_aspect('equal')
ax.set_xlim(7.5, 12.5)
ax.set_ylim(6.5, 10.5)
#ax.invert_yaxis()  # Optional: standard nuclear chart layout (Z increases downward)

plt.tight_layout()
plt.savefig("nuclear_chart.png", dpi=300, bbox_inches="tight") 
plt.show()
```
