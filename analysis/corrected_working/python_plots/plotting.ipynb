```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
```

```python
def load_sf(filename):
    data = np.loadtxt(filename)
    energy = data[:, 0]
    ell0   = data[:, 1]
    ell2   = data[:, 2]
    unc0   = data[:, 3]
    unc2   = data[:, 4]

    idx = np.argsort(energy)
    return energy[idx], ell0[idx], ell2[idx], unc0[idx], unc2[idx]
```

```python
def load_sf_shift(filename):
    data = np.loadtxt(filename)
    energy = data[:, 0] + 1.08
    ell0   = data[:, 1]
    ell2   = data[:, 2]
    unc0   = data[:, 3]
    unc2   = data[:, 4]

    idx = np.argsort(energy)
    return energy[idx], ell0[idx], ell2[idx], unc0[idx], unc2[idx]
```

```python
def vertical_mirror_plot(top_data, bottom_data, top_title, bottom_title, main_title):

    ET, e0T, e2T, u0T, u2T = top_data
    EB, e0B, e2B, u0B, u2B = bottom_data

    bar_width = 0.08
    
    fig = plt.figure(figsize=(7,6))
    gs = fig.add_gridspec(2, 1, height_ratios=[1,1], hspace=0.002)

    axT = fig.add_subplot(gs[0,0])
    axB = fig.add_subplot(gs[1,0], sharex=axT)

    axT.bar(ET, e0T, width=bar_width, yerr=u0T,
            color='white', edgecolor='tab:blue', hatch='\\', fill=True, capsize=3, linewidth=2)
    axT.bar(ET, e2T, width=bar_width, bottom=e0T, yerr=u2T,
            color='tab:orange', edgecolor='tab:orange', capsize=3, linewidth=2)
    axT.set_ylim(0, 3)
    axT.set_ylabel("Spectroscopic Factor", fontsize=11)
    axT.tick_params(axis='x', which='both', bottom=False, top=False, labelbottom=False, direction='in')
    axT.tick_params(axis='y', direction='in')
    yticks = axT.get_yticks()
    axT.set_yticks(yticks[:-1])

    axT.text(x=0.5, y=2.5, s=top_title, fontsize=14, color='black')

    axB.bar(EB, e0B, width=bar_width, yerr=u0B,
            color='white', edgecolor='tab:blue', hatch='\\', fill=True, capsize=3, linewidth=2, label=r'$\ell=0$')
    axB.bar(EB, e2B, width=bar_width, bottom=e0B, yerr=u2B,
            color='tab:orange', edgecolor='tab:orange', capsize=3, label=r'$\ell=2$')
    axB.set_ylabel("Spectroscopic Factor", fontsize=11)
    axB.set_ylim(0, 3)
    axB.tick_params(axis='x', direction='in')
    axB.tick_params(axis='y', direction='in')
    yticks = axB.get_yticks()
    axB.set_yticks(yticks[:-1])
    axB.text(x=0.5, y=2.5, s=bottom_title, fontsize=14, color='black')
    axB.legend(loc="upper right", fontsize=10)

    axB.set_xlabel("Excitation Energy (MeV)", fontsize=11)

    fig.suptitle(main_title, fontsize=18)
    fig.patch.set_facecolor('white')

    plt.show()
```

```python
def mirror_plot(left_data, right_data, left_title, right_title, main_title, title):
    
    EL, e0L, e2L, u0L, u2L = left_data
    ER, e0R, e2R, u0R, u2R = right_data

    bar_height = 0.08

    fig = plt.figure(figsize=(11, 7))
    gs = fig.add_gridspec(1, 3, width_ratios=[1.2, 0.15, 1.2], wspace=0.0)

    axL = fig.add_subplot(gs[0, 0])
    axC = fig.add_subplot(gs[0, 1], sharey=axL)
    axR = fig.add_subplot(gs[0, 2], sharey=axL)
    
    fig.subplots_adjust(wspace=0)
    
    axL.barh(EL, -e0L, height=bar_height,
            xerr=u0L, color='white', edgecolor='tab:blue', hatch='\\', fill=True, capsize=3, linewidth=2)

    axL.barh(EL, -e2L, height=bar_height, left=-e0L,
             xerr=u2L, color='tab:orange', capsize=3, linewidth=2)

    axL.set_xlim(-3, 0)
    axL.set_xlabel("Spectroscopic Factor", fontsize=11)
    axL.tick_params(axis='y', left=False, labelleft=False)
    axL.set_title(left_title, fontsize=12)
        
    axR.barh(ER, e0R, height=bar_height,
            xerr=u0R,
            color='white', edgecolor='tab:blue', hatch='\\', fill=True, linewidth=2,
            label=r'$\ell=0$', capsize=3)

    axR.barh(ER, e2R, height=bar_height, left=e0R,
             xerr=u2R, color='tab:orange',
             label=r'$\ell=2$', capsize=3, linewidth=2)
    
    axR.set_xlim(0, 3)
    axR.set_xlabel("Spectroscopic Factor", fontsize=11)
    axR.tick_params(which='both', right=False, labelleft=False, labelright=False)
    axR.set_title(right_title, fontsize=12)
    axR.legend(loc="upper right", fontsize=10)
    
    #for spine in ['top', 'right', 'left', 'bottom']:
     #   axR.spines[spine].set_visible(False)
        
    axR.legend(loc="upper right", fontsize=10)
    
    axC.set_xlim(0, 1)
    axC.set_xticks([])

    for spine in ['top', 'bottom', 'right']:
        axC.spines[spine].set_visible(False)

    axC.spines['left'].set_visible(True)
    axC.spines['left'].set_linewidth(1.2)

    axC.yaxis.set_label_position("left")
    axC.yaxis.tick_left()

    axC.set_ylabel("Excitation Energy (MeV)",
                   fontsize=13,
                   labelpad=14)

    axC.yaxis.set_label_coords(0.58, 0.5)

    axC.tick_params(axis='y',
                    which='both',
                    direction='in',
                    pad=4,
                    width=1,
                    length=6,
                    left=True,
                    right=False,
                    labelleft=True)
    
    axL.spines['right'].set_visible(True)
    axR.spines['left'].set_visible(True)

    axC.spines['left'].set_position(('axes', 1.0))
    axC.spines['right'].set_position(('axes', 1.0))

    fig.suptitle(main_title, fontsize=14)
    fig.patch.set_facecolor('white')
    
    plt.rcParams['axes.linewidth'] = 1.2

    plt.savefig(title, dpi=600, bbox_inches="tight", format="pdf")
    plt.show()
```

```python
def mirror_plot_4(left1, right2, left3, right4, main_title, filename):
    datasets = [left1, right2, left3, right4]
    titles = [r"$^{18}$O, SM", r"$^{18}$O, Experimental Data", r"$^{18}$F, Experimental Data", r"$^{18}$F, SM"]
    directions = ["left", "right", "left", "right"]

    bar_height = 0.08
    
    fig = plt.figure(figsize=(22, 7))
    gs = fig.add_gridspec(1, 5, width_ratios=[1.2, 1.2, 0.2, 1.2, 1.2], wspace=0.0)

    ax0 = fig.add_subplot(gs[0, 0])
    ax1 = fig.add_subplot(gs[0, 1], sharey=ax0)
    axC = fig.add_subplot(gs[0, 2], sharey=ax0)
    ax2 = fig.add_subplot(gs[0, 3], sharey=ax0)
    ax3 = fig.add_subplot(gs[0, 4], sharey=ax0)
    
    axes = [ax0, ax1, ax2, ax3]

    for i, (data, direction) in enumerate(zip(datasets, directions)):
        E, e0, e2, u0, u2 = data
        ax = axes[i]

        if direction == "left":
            ax.barh(E, e0, height=bar_height, xerr=u0,
                    color='white', edgecolor='tab:blue', hatch='\\', fill=True, capsize=3, linewidth=2)
            ax.barh(E, e2, height=bar_height, left=e0, xerr=u2,
                    color='tab:orange', capsize=3, linewidth=2)
            ax.set_xlim(0, 2.99)
        else:
            ax.barh(E, -e0, height=bar_height, xerr=u0,
                    color='white', edgecolor='tab:blue', hatch='\\', fill=True, capsize=3, linewidth=2)
            ax.barh(E, -e2, height=bar_height, left=-e0, xerr=u2,
                    color='tab:orange', capsize=3, linewidth=2)
            ax.set_xlim(-2.99, 0)

        ax.set_title(titles[i], fontsize=12)
        ax.tick_params(axis='y', left=False, labelleft=False)
        ax.set_xlabel("Spectroscopic Factor", fontsize=11)

        if i == 0:
            ax.legend([r'$\ell=0$', r'$\ell=2$'], loc="upper right", fontsize=10, frameon=False)
    
    for ax in axes:
        ax.tick_params(axis='y', which='both', bottom=False, top=False, labelbottom=False)

        axes[3].tick_params(axis='y', which='both', bottom=True, top=False, labelbottom=True)
    
    axC.set_xlim(0, 1)
    axC.set_xticks([])

    for spine in ['top', 'bottom', 'right']:
        axC.spines[spine].set_visible(False)

    axC.spines['left'].set_visible(True)
    axC.spines['left'].set_linewidth(1.2)

    axC.yaxis.set_label_position("left")
    axC.yaxis.tick_left()

    axC.set_ylabel("Excitation Energy (MeV)",
                   fontsize=13,
                   labelpad=14)

    axC.yaxis.set_label_coords(0.58, 0.5)

    axC.tick_params(axis='y',
                    which='both',
                    direction='out',
                    pad=4,
                    width=1,
                    length=6,
                    left=True,
                    right=True,
                    labelleft=True)
    
    ax1.spines['right'].set_visible(True)
    ax2.spines['left'].set_visible(True)

    axC.spines['left'].set_position(('axes', 1.0))
    axC.spines['right'].set_position(('axes', 1.0))
    
    fig.suptitle(main_title, fontsize=14)
    fig.patch.set_facecolor('white')
    
    #axes[0].spines['right'].set_visible(False)
    #axes[1].spines['left'].set_visible(False)
    #axes[2].spines['right'].set_visible(False)
    #axes[3].spines['left'].set_visible(False)

    plt.rcParams['axes.linewidth'] = 1.2
    plt.savefig(filename, dpi=600, bbox_inches="tight", format="pdf")
    plt.show()
    
mirror_plot_4(
    O_SM_shifted, O_exp_shifted, F_exp, F_SM,
    r"Spectroscopic Factors", "18O_18F_SM.pdf"
)
```

```python
def mirror_plot_4_2yax(left1, right2, left3, right4, summed_left, summed_right, main_title, filename):
    datasets = [left1, right2, left3, right4]
    titles = [r"$^{18}$O, SM", r"$^{18}$O, Experimental Data", r"$^{18}$F, Experimental Data", r"$^{18}$F, SM"]
    directions = ["left", "right", "left", "right"]

    bar_height = 0.08
    sum_bar_height = np.array([0.6, 1.2, 1.2])
    
    fig = plt.figure(figsize=(22, 7))
    gs = fig.add_gridspec(1, 4, width_ratios=[1., 1., 1., 1.], wspace=0.0)

    ax0 = fig.add_subplot(gs[0, 0])
    ax1 = fig.add_subplot(gs[0, 1], sharey=ax0)
    ax2 = fig.add_subplot(gs[0, 2], sharey=ax0)
    ax3 = fig.add_subplot(gs[0, 3], sharey=ax0)
    
    axes = [ax0, ax1, ax2, ax3]
    
    E, e0, e2, u0, u2 = summed_left
    ax0.barh(E, e0, height=bar_height,
             color='tab:blue', alpha=0.2, edgecolor='none', zorder=0)
    ax0.barh(E, e2, height=bar_height, left=e0,
             color='tab:orange', alpha=0.2, edgecolor='none', zorder=0)

    E, e0, e2, u0, u2 = summed_right
    ax3.barh(E, -e0, height=sum_bar_height,
             color='tab:blue', alpha=0.2, edgecolor='none', zorder=0)
    ax3.barh(E, -e2, height=sum_bar_height, left=-e0,
             color='tab:orange', alpha=0.2, edgecolor='none', zorder=0)

    for i, (data, direction) in enumerate(zip(datasets, directions)):
        E, e0, e2, u0, u2 = data
        ax = axes[i]

        if direction == "left":
            ax.barh(E, e0, height=bar_height, xerr=u0,
                    color='tab:blue', edgecolor='black', hatch='\\', fill=True, capsize=3, linewidth=1)
            ax.barh(E, e2, height=bar_height, left=e0, xerr=u2,
                    color='tab:orange', edgecolor='black', capsize=3, linewidth=1)
            ax.set_xlim(0, 2.99)

        else:
            ax.barh(E, -e0, height=bar_height, xerr=u0,
                    color='tab:blue', edgecolor='black', hatch='\\', fill=True, capsize=3, linewidth=1)
            ax.barh(E, -e2, height=bar_height, left=-e0, xerr=u2,
                    color='tab:orange', edgecolor='black', capsize=3, linewidth=1)
            ax.set_xlim(-2.99, 0)

        ax.set_title(titles[i], fontsize=12)
        ax.tick_params(axis='y', left=False, labelleft=False)
        ax.set_xlabel("Spectroscopic Factor", fontsize=11)
        
        legend_handles = [
            Patch(facecolor='tab:blue', edgecolor='black', hatch='\\', linewidth=1, label=r'$\ell=0$'),
            Patch(facecolor='tab:orange', edgecolor='black', linewidth=1, label=r'$\ell=2$')
        ]

        if i == 0:
            ax.legend(handles=legend_handles, loc="upper right", fontsize=10, frameon=False)
    
    for ax in axes:
        ax.tick_params(axis='y', which='both',
                   left=False, right=False,
                   labelleft=False, labelright=False)

    axes[0].set_ylabel("Excitation Energy (MeV)", fontsize=13, labelpad=14)
    axes[0].tick_params(axis='y',
                    which='both',
                    direction='out',
                    pad=4,
                    width=1,
                    length=6,
                    left=True,
                    right=False,
                    labelleft=True)

    axes[0].spines['left'].set_visible(True)
    axes[0].spines['right'].set_visible(False)

    axes[3].set_ylabel("Excitation Energy (MeV)", fontsize=13, labelpad=14)
    axes[3].yaxis.set_label_position("right")

    axes[3].tick_params(axis='y',
                    which='both',
                    direction='out',
                    pad=4,
                    width=1,
                    length=6,
                    left=False,
                    right=True,
                    labelright=True)

    axes[3].spines['right'].set_visible(True)
    axes[3].spines['left'].set_visible(False)
    
    fig.suptitle(main_title, fontsize=14)
    fig.patch.set_facecolor('white')
    
    #axes[0].spines['right'].set_visible(False)
    #axes[1].spines['left'].set_visible(False)
    #axes[2].spines['right'].set_visible(False)
    #axes[3].spines['left'].set_visible(False)

    plt.rcParams['axes.linewidth'] = 1.2
    plt.savefig(filename, dpi=600, bbox_inches="tight", format="pdf")
    plt.show()
    
mirror_plot_4_2yax(
    O_SMi_shifted, O_exp_shifted, F_exp, F_SMi, O_SMi_summed_shifted, F_SMi_summed,
    r"Spectroscopic Factors, USDi", "18O_18F_4_2yax_usdi.pdf"
)
```

```python
def mirror_plot_4_2yax(left1, right2, left3, right4, summed_left, summed_right, main_title, filename):
    datasets = [left1, right2, left3, right4]
    titles = [r"$^{18}$O, SM", r"$^{18}$O, Experimental Data", r"$^{18}$F, Experimental Data", r"$^{18}$F, SM"]
    directions = ["left", "right", "left", "right"]

    bar_height = 0.08
    sum_bar_height = np.array([0.8, 1.05])
    
    fig = plt.figure(figsize=(22, 7))
    gs = fig.add_gridspec(1, 4, width_ratios=[1., 1., 1., 1.], wspace=0.0)

    ax0 = fig.add_subplot(gs[0, 0])
    ax1 = fig.add_subplot(gs[0, 1], sharey=ax0)
    ax2 = fig.add_subplot(gs[0, 2], sharey=ax0)
    ax3 = fig.add_subplot(gs[0, 3], sharey=ax0)
    
    axes = [ax0, ax1, ax2, ax3]
    
    E, e0, e2, u0, u2 = summed_left
    ax0.barh(E, e0, height=bar_height,
             color='tab:blue', alpha=0.2, edgecolor='none', zorder=0)
    ax0.barh(E, e2, height=bar_height, left=e0,
             color='tab:orange', alpha=0.2, edgecolor='none', zorder=0)

    E, e0, e2, u0, u2 = summed_right
    ax3.barh(E, -e0, height=sum_bar_height,
             color='tab:blue', alpha=0.2, edgecolor='none', zorder=0)
    ax3.barh(E, -e2, height=sum_bar_height, left=-e0,
             color='tab:orange', alpha=0.2, edgecolor='none', zorder=0)

    for i, (data, direction) in enumerate(zip(datasets, directions)):
        E, e0, e2, u0, u2 = data
        ax = axes[i]

        if direction == "left":
            ax.barh(E, e0, height=bar_height, xerr=u0,
                    color='tab:blue', edgecolor='black', hatch='\\', fill=True, capsize=3, linewidth=1)
            ax.barh(E, e2, height=bar_height, left=e0, xerr=u2,
                    color='tab:orange', edgecolor='black', capsize=3, linewidth=1)
            ax.set_xlim(0, 2.99)

        else:
            ax.barh(E, -e0, height=bar_height, xerr=u0,
                    color='tab:blue', edgecolor='black', hatch='\\', fill=True, capsize=3, linewidth=1)
            ax.barh(E, -e2, height=bar_height, left=-e0, xerr=u2,
                    color='tab:orange', edgecolor='black', capsize=3, linewidth=1)
            ax.set_xlim(-2.99, 0)

        ax.set_title(titles[i], fontsize=12)
        ax.tick_params(axis='y', left=False, labelleft=False)
        ax.set_xlabel("Spectroscopic Factor", fontsize=11)
        
        legend_handles = [
            Patch(facecolor='tab:blue', edgecolor='black', hatch='\\', linewidth=1, label=r'$\ell=0$'),
            Patch(facecolor='tab:orange', edgecolor='black', linewidth=1, label=r'$\ell=2$')
        ]

        if i == 0:
            ax.legend(handles=legend_handles, loc="upper right", fontsize=10, frameon=False)
    
    for ax in axes:
        ax.tick_params(axis='y', which='both',
                   left=False, right=False,
                   labelleft=False, labelright=False)

    axes[0].set_ylabel("Excitation Energy (MeV)", fontsize=13, labelpad=14)
    axes[0].tick_params(axis='y',
                    which='both',
                    direction='out',
                    pad=4,
                    width=1,
                    length=6,
                    left=True,
                    right=False,
                    labelleft=True)

    axes[0].spines['left'].set_visible(True)
    axes[0].spines['right'].set_visible(False)

    axes[3].set_ylabel("Excitation Energy (MeV)", fontsize=13, labelpad=14)
    axes[3].yaxis.set_label_position("right")

    axes[3].tick_params(axis='y',
                    which='both',
                    direction='out',
                    pad=4,
                    width=1,
                    length=6,
                    left=False,
                    right=True,
                    labelright=True)

    axes[3].spines['right'].set_visible(True)
    axes[3].spines['left'].set_visible(False)
    
    fig.suptitle(main_title, fontsize=14)
    fig.patch.set_facecolor('white')
    
    #axes[0].spines['right'].set_visible(False)
    #axes[1].spines['left'].set_visible(False)
    #axes[2].spines['right'].set_visible(False)
    #axes[3].spines['left'].set_visible(False)

    plt.rcParams['axes.linewidth'] = 1.2
    plt.savefig(filename, dpi=600, bbox_inches="tight", format="pdf")
    plt.show()

mirror_plot_4_2yax(
    O_SMfsu_shifted, O_exp_shifted, F_exp, F_SMfsu, O_SMfsu_summed_shifted, F_SMfsu_summed,
    r"Spectroscopic Factors, FSU", "18O_18F_4_2yax_fsu.pdf"
)
```

```python
def mirror_plot_4_2yax_lines(
    left1, right2, left3, right4, summed_left, summed_right,
    main_title, filename,
    dashed_lines=None,
    annotations=None
):
    datasets = [left1, right2, left3, right4]
    titles = [r"$^{18}$O, SM", r"$^{18}$O, Experimental Data",
              r"$^{18}$F, Experimental Data", r"$^{18}$F, SM"]
    directions = ["left", "right", "left", "right"]
    
    sum_bar_height = np.array([0.6, 1.0, 1.0])

    fig = plt.figure(figsize=(22, 7))
    gs = fig.add_gridspec(1, 4, width_ratios=[1., 1., 1., 1.], wspace=0.0)

    ax0 = fig.add_subplot(gs[0, 0])
    ax1 = fig.add_subplot(gs[0, 1], sharey=ax0)
    ax2 = fig.add_subplot(gs[0, 2], sharey=ax0)
    ax3 = fig.add_subplot(gs[0, 3], sharey=ax0)

    axes = [ax0, ax1, ax2, ax3]

    E, e0, e2, u0, u2 = summed_left
    ax0.barh(E, e0, height=bar_height, color='tab:blue', alpha=0.2, edgecolor='none', zorder=0)
    ax0.barh(E, e2, height=bar_height, left=e0, color='tab:orange', alpha=0.2, edgecolor='none', zorder=0)

    E, e0, e2, u0, u2 = summed_right
    ax3.barh(E, -e0, height=sum_bar_height, color='tab:blue', alpha=0.2, edgecolor='none', zorder=0)
    ax3.barh(E, -e2, height=sum_bar_height, left=-e0, color='tab:orange', alpha=0.2, edgecolor='none', zorder=0)

    for i, (data, direction) in enumerate(zip(datasets, directions)):
        E, e0, e2, u0, u2 = data
        ax = axes[i]

        if direction == "left":
            height_array = bar_height
        else:
            height_array = bar_height

        if direction == "left":
            ax.barh(E, e0, height=height_array, xerr=u0,
                    color='tab:blue', edgecolor='black', hatch='\\', fill=True, capsize=3, linewidth=1)
            ax.barh(E, e2, height=height_array, left=e0, xerr=u2,
                    color='tab:orange', edgecolor='black', capsize=3, linewidth=1)
            ax.set_xlim(0, 2.99)
        else:
            ax.barh(E, -e0, height=height_array, xerr=u0,
                    color='tab:blue', edgecolor='black', hatch='\\', fill=True, capsize=3, linewidth=1)
            ax.barh(E, -e2, height=height_array, left=-e0, xerr=u2,
                    color='tab:orange', edgecolor='black', capsize=3, linewidth=1)
            ax.set_xlim(-2.99, 0)

        ax.set_title(titles[i], fontsize=15)
        ax.tick_params(axis='y', left=False, labelleft=False)
        ax.set_xlabel("Spectroscopic Factor", fontsize=15, labelpad=14)

        if i == 0:
            legend_handles = [
                Patch(facecolor='tab:blue', edgecolor='black', hatch='\\', linewidth=1, label=r'$\ell=0$'),
                Patch(facecolor='tab:orange', edgecolor='black', linewidth=1, label=r'$\ell=2$')
            ]
            ax.legend(handles=legend_handles, loc="lower right", fontsize=14, frameon=False)

        if annotations and i < len(annotations):
            for idx, (text, x_ann, y_ann) in annotations[i].items():
                ax.text(x_ann, y_ann, text, fontsize=12, ha='center', va='bottom')

    for ax_start_idx, ax_end_idx, x_start, y_start, x_end, y_end in dashed_lines:
        ax_start = axes[ax_start_idx]
        ax_end = axes[ax_end_idx]

        start_fig = ax_start.transData.transform((x_start, y_start))
        end_fig = ax_end.transData.transform((x_end, y_end))

        start_fig_norm = fig.transFigure.inverted().transform(start_fig)
        end_fig_norm = fig.transFigure.inverted().transform(end_fig)

        line = Line2D([start_fig_norm[0], end_fig_norm[0]],
                      [start_fig_norm[1], end_fig_norm[1]],
                      linestyle='--', color='k', linewidth=1,
                      transform=fig.transFigure, clip_on=False)
        fig.add_artist(line)

    for ax in axes:
        ax.tick_params(axis='y', which='both', left=False, right=False, labelleft=False, labelright=False)
        ax.tick_params(axis='x', labelsize=14)

    axes[0].set_ylabel("Excitation Energy (MeV)", fontsize=15, labelpad=14)
    axes[0].tick_params(axis='y', which='both', direction='out', pad=4, width=1, length=6, left=True, right=False, labelleft=True, labelsize=14)
    axes[0].spines['left'].set_visible(True)
    axes[0].spines['right'].set_visible(False)
    axes[1].spines['left'].set_visible(False)

    axes[3].set_ylabel("Excitation Energy (MeV)", fontsize=15, labelpad=14)
    axes[3].yaxis.set_label_position("right")
    axes[3].tick_params(axis='y', which='both', direction='out', pad=4, width=1, length=6, left=False, right=True, labelright=True, labelsize=14)
    axes[3].spines['right'].set_visible(True)
    axes[3].spines['left'].set_visible(False)
    axes[2].spines['right'].set_visible(False)

    fig.suptitle(main_title, fontsize=16)
    fig.patch.set_facecolor('white')
    plt.rcParams['axes.linewidth'] = 1.2
    plt.savefig(filename, dpi=600, bbox_inches="tight", format="pdf")
    plt.show()
```
