```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
energy = np.array([
    6.1495, 5.8995, 5.6495, 5.3995, 5.1495,
    4.8995, 4.6495, 4.3995, 4.1495, 3.8995, 3.6495
])

d_vals = np.array([
    1.0202, 1.02495, 1.0289, 0.41636, 0.5823,
    0.75, 0.92, 1.0826, 1.25, 1.42092, 1.59003
])

s_vals = np.array([
    2.63755, 2.65, 2.66726, 1.23114, 1.169,
    1.112, 1.058, 1.01067, 0.96, 0.93427, 0.9026
])

unc_d = np.array([
    0.27, 0.27, 0.27, 0.51, 0.52,
    0.52, 0.55, 0.58, 0.58, 0.58, 0.61
])

unc_s = np.array([
    0.27, 0.27, 0.28, 0.13, 0.13,
    0.12, 0.11, 0.10, 0.10, 0.10, 0.09
])
```

```python
energy_fit = energy[3:]
d_fit = d_vals[3:]
s_fit = s_vals[3:]
unc_d_fit = unc_d[3:]
unc_s_fit = unc_s[3:]
```

```python
def plo2(E, a, b, c):
    return a + b*E + c*E**2
```

```python
popt_d, pcov_d = curve_fit(plo2, energy_fit, d_fit)
popt_s, pcov_s = curve_fit(plo2, energy_fit, s_fit)

print("Fit parameters for 0d5/2:")
print("a =", popt_d[0])
print("b =", popt_d[1])
print("c =", popt_d[2])

print("\nFit parameters for 1s1/2:")
print("a =", popt_s[0])
print("b =", popt_s[1])
print("c =", popt_s[2])
```

```python
E_plot = np.linspace(min(energy_fit), 6.5, 300)

plt.plot(E_plot, plo2(E_plot, *popt_d), '-', label='0d5/2 fit')

plt.plot(E_plot, plo2(E_plot, *popt_s), '--', label='1s1/2 fit')

plt.scatter(energy, d_vals, color='red')
plt.scatter(energy, s_vals, color='black')

plt.legend()

plt.xlabel("Energy (MeV)")
plt.ylabel("SF")
```
